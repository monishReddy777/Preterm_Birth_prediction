```python
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch_geometric.nn import SAGEConv
from torch_geometric.data import Data
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import classification_report, confusion_matrix, roc_curve, auc, precision_recall_curve, accuracy_score
import numpy as np
import pandas as pd
import seaborn as sns
from sklearn.preprocessing import RobustScaler
from imblearn.combine import SMOTETomek
import matplotlib.pyplot as plt

# Load and preprocess data
df1 = pd.read_csv("final_features_tpehg.csv")
df2 = pd.read_csv("features_toco.csv")

df1.drop(columns=["RecID", "Hypertension", "Diabetes", "Bleeding_first_trimester", "Bleeding_second_trimester", "Funneling", "Early"], errors='ignore', inplace=True)
df2.drop(columns=["Record"], errors='ignore', inplace=True)

df = pd.concat([df1, df2], axis=0).reset_index(drop=True)

X = df.drop(columns=["Premature"])
y = df["Premature"]

scaler = RobustScaler()
X_scaled = scaler.fit_transform(X)

class EnhancedGraphSAGEModel(nn.Module):
    def __init__(self, in_channels, hidden_channels=64, out_channels=2, dropout=0.3):
        super(EnhancedGraphSAGEModel, self).__init__()
        self.fc_identity = nn.Linear(in_channels, hidden_channels)
        self.conv1 = SAGEConv(in_channels, hidden_channels, aggr="mean")
        self.conv2 = SAGEConv(hidden_channels, hidden_channels, aggr="max")
        self.conv3 = SAGEConv(hidden_channels, hidden_channels, aggr="mean")
        self.fc1 = nn.Linear(hidden_channels, hidden_channels // 2)
        self.fc2 = nn.Linear(hidden_channels // 2, out_channels)
        self.dropout = nn.Dropout(dropout)
        
    def forward(self, x, edge_index):
        identity = self.fc_identity(x)
        x = self.conv1(x, edge_index)
        x = F.elu(x)
        x = self.dropout(x)
        x = self.conv2(x, edge_index) + identity
        x = F.elu(x)
        x = self.dropout(x)
        x = self.conv3(x, edge_index)
        x = F.elu(x)
        x = self.dropout(x)
        x = self.fc1(x)
        x = F.elu(x)
        x = self.dropout(x)
        x = self.fc2(x)
        return x

def train_model(model, train_data, optimizer, criterion):
    model.train()
    optimizer.zero_grad()
    out = model(train_data.x, train_data.edge_index)
    loss = criterion(out, train_data.y)
    loss.backward()
    optimizer.step()
    return loss.item()

def evaluate_model(model, data):
    model.eval()
    with torch.no_grad():
        out = model(data.x, data.edge_index)
        return out.argmax(dim=1), out.softmax(dim=1)[:, 1]

skf = StratifiedKFold(n_splits=10, shuffle=True, random_state=42)
overall_accuracy, overall_sensitivity, overall_specificity = [], [], []
train_losses, all_y_true, all_y_scores = [], [], []

for fold, (train_idx, val_idx) in enumerate(skf.split(X_scaled, y)):
    print(f"\nFold {fold+1}")
    X_train, X_val = X_scaled[train_idx], X_scaled[val_idx]
    y_train, y_val = y.iloc[train_idx], y.iloc[val_idx]
    
    smote_tomek = SMOTETomek(sampling_strategy='minority', random_state=42)
    X_train_resampled, y_train_resampled = smote_tomek.fit_resample(X_train, y_train)
    
    train_data = Data(x=torch.tensor(X_train_resampled, dtype=torch.float), edge_index=torch.randint(0, X_train_resampled.shape[0], (2, X_train_resampled.shape[0] * 2)), y=torch.tensor(y_train_resampled.to_numpy(), dtype=torch.long))
    val_data = Data(x=torch.tensor(X_val, dtype=torch.float), edge_index=torch.randint(0, X_val.shape[0], (2, X_val.shape[0] * 2)), y=torch.tensor(y_val.to_numpy(), dtype=torch.long))
    
    model = EnhancedGraphSAGEModel(in_channels=X_train_resampled.shape[1], hidden_channels=64, out_channels=2)
    optimizer = torch.optim.AdamW(model.parameters(), lr=0.001, weight_decay=1e-4)
    criterion = nn.CrossEntropyLoss()
    
    fold_losses = []
    for epoch in range(150):
        loss = train_model(model, train_data, optimizer, criterion)
        fold_losses.append(loss)
    train_losses.append(fold_losses)
    
    val_preds, val_scores = evaluate_model(model, val_data)
    all_y_true.extend(y_val)
    all_y_scores.extend(val_scores.numpy())
    
    acc = accuracy_score(y_val, val_preds)
    cm = confusion_matrix(y_val, val_preds)
    tn, fp, fn, tp = cm.ravel()
    sensitivity = tp / (tp + fn)
    specificity = tn / (tn + fp)
    
    overall_accuracy.append(acc)
    overall_sensitivity.append(sensitivity)
    overall_specificity.append(specificity)
    
    print(classification_report(y_val, val_preds))

print("\nFinal 10-Fold Results:")
print(f"Mean Accuracy: {np.mean(overall_accuracy):.4f} ± {np.std(overall_accuracy):.4f}")
print(f"Mean Sensitivity: {np.mean(overall_sensitivity):.4f} ± {np.std(overall_sensitivity):.4f}")
print(f"Mean Specificity: {np.mean(overall_specificity):.4f} ± {np.std(overall_specificity):.4f}")

plt.figure()
for i, losses in enumerate(train_losses):
    plt.plot(losses, label=f'Fold {i+1}')
plt.legend()
plt.title("Training Loss Across Folds")
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.show()

fpr, tpr, _ = roc_curve(all_y_true, all_y_scores)
plt.plot(fpr, tpr, label=f'AUC: {auc(fpr, tpr):.3f}')
plt.legend()
plt.title("AUC-ROC Curve")
plt.show()

precision, recall, _ = precision_recall_curve(all_y_true, all_y_scores)
plt.plot(recall, precision, label="Precision-Recall Curve")
plt.legend()
plt.title("Precision-Recall Curve")
plt.show()
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch_geometric.nn import SAGEConv
from torch_geometric.data import Data
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import classification_report, confusion_matrix, roc_curve, auc, precision_recall_curve, accuracy_score
import numpy as np
import pandas as pd
import seaborn as sns
from sklearn.preprocessing import RobustScaler
from imblearn.combine import SMOTETomek
import matplotlib.pyplot as plt

# Load and preprocess data
df1 = pd.read_csv("final_features_tpehg.csv")
df2 = pd.read_csv("features_toco.csv")

df1.drop(columns=["RecID", "Hypertension", "Diabetes", "Bleeding_first_trimester", "Bleeding_second_trimester", "Funneling", "Early"], errors='ignore', inplace=True)
df2.drop(columns=["Record"], errors='ignore', inplace=True)

df = pd.concat([df1, df2], axis=0).reset_index(drop=True)

X = df.drop(columns=["Premature"])
y = df["Premature"]

scaler = RobustScaler()
X_scaled = scaler.fit_transform(X)

class EnhancedGraphSAGEModel(nn.Module):
    def __init__(self, in_channels, hidden_channels=64, out_channels=2, dropout=0.3):
        super(EnhancedGraphSAGEModel, self).__init__()
        self.fc_identity = nn.Linear(in_channels, hidden_channels)
        self.conv1 = SAGEConv(in_channels, hidden_channels, aggr="mean")
        self.conv2 = SAGEConv(hidden_channels, hidden_channels, aggr="max")
        self.conv3 = SAGEConv(hidden_channels, hidden_channels, aggr="mean")
        self.fc1 = nn.Linear(hidden_channels, hidden_channels // 2)
        self.fc2 = nn.Linear(hidden_channels // 2, out_channels)
        self.dropout = nn.Dropout(dropout)
        
    def forward(self, x, edge_index):
        identity = self.fc_identity(x)
        x = self.conv1(x, edge_index)
        x = F.elu(x)
        x = self.dropout(x)
        x = self.conv2(x, edge_index) + identity
        x = F.elu(x)
        x = self.dropout(x)
        x = self.conv3(x, edge_index)
        x = F.elu(x)
        x = self.dropout(x)
        x = self.fc1(x)
        x = F.elu(x)
        x = self.dropout(x)
        x = self.fc2(x)
        return x

def train_model(model, train_data, optimizer, criterion):
    model.train()
    optimizer.zero_grad()
    out = model(train_data.x, train_data.edge_index)
    loss = criterion(out, train_data.y)
    loss.backward()
    optimizer.step()
    return loss.item(), accuracy_score(train_data.y.numpy(), out.argmax(dim=1).numpy())

def evaluate_model(model, data):
    model.eval()
    with torch.no_grad():
        out = model(data.x, data.edge_index)
        return out.argmax(dim=1), out.softmax(dim=1)[:, 1]

skf = StratifiedKFold(n_splits=10, shuffle=True, random_state=42)
overall_accuracy, overall_sensitivity, overall_specificity = [], [], []
train_losses, train_accuracies, all_y_true, all_y_scores = [], [], [], []

for fold, (train_idx, val_idx) in enumerate(skf.split(X_scaled, y)):
    print(f"\nFold {fold+1}")
    X_train, X_val = X_scaled[train_idx], X_scaled[val_idx]
    y_train, y_val = y.iloc[train_idx], y.iloc[val_idx]
    
    smote_tomek = SMOTETomek(sampling_strategy='minority', random_state=42)
    X_train_resampled, y_train_resampled = smote_tomek.fit_resample(X_train, y_train)
    
    train_data = Data(x=torch.tensor(X_train_resampled, dtype=torch.float), edge_index=torch.randint(0, X_train_resampled.shape[0], (2, X_train_resampled.shape[0] * 2)), y=torch.tensor(y_train_resampled.to_numpy(), dtype=torch.long))
    val_data = Data(x=torch.tensor(X_val, dtype=torch.float), edge_index=torch.randint(0, X_val.shape[0], (2, X_val.shape[0] * 2)), y=torch.tensor(y_val.to_numpy(), dtype=torch.long))
    
    model = EnhancedGraphSAGEModel(in_channels=X_train_resampled.shape[1])
    optimizer = torch.optim.AdamW(model.parameters(), lr=0.001, weight_decay=1e-4)
    criterion = nn.CrossEntropyLoss()
    
    fold_losses, fold_accuracies = [], []
    for epoch in range(150):
        loss, acc = train_model(model, train_data, optimizer, criterion)
        fold_losses.append(loss)
        fold_accuracies.append(acc)
    train_losses.append(fold_losses)
    train_accuracies.append(fold_accuracies)
    
    val_preds, val_scores = evaluate_model(model, val_data)
    all_y_true.extend(y_val)
    all_y_scores.extend(val_scores.numpy())
    
    acc = accuracy_score(y_val, val_preds)
    cm = confusion_matrix(y_val, val_preds)
    tn, fp, fn, tp = cm.ravel()
    sensitivity = tp / (tp + fn)
    specificity = tn / (tn + fp)
    
    overall_accuracy.append(acc)
    overall_sensitivity.append(sensitivity)
    overall_specificity.append(specificity)
    
    print(classification_report(y_val, val_preds))

print("\nFinal 10-Fold Results:")
print(f"Mean Accuracy: {np.mean(overall_accuracy):.4f} ± {np.std(overall_accuracy):.4f}")
print(f"Mean Sensitivity: {np.mean(overall_sensitivity):.4f} ± {np.std(overall_sensitivity):.4f}")
print(f"Mean Specificity: {np.mean(overall_specificity):.4f} ± {np.std(overall_specificity):.4f}")

plt.figure()
for i, losses in enumerate(train_losses):
    plt.plot(losses, label=f'Fold {i+1}')
plt.legend()
plt.title("Training Loss Across Folds")
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.show()

plt.figure()
for i, acc in enumerate(train_accuracies):
    plt.plot(acc, label=f'Fold {i+1}')
plt.legend()
plt.title("Training Accuracy Across Folds")
plt.xlabel("Epochs")
plt.ylabel("Accuracy")
plt.show()

cm = confusion_matrix(all_y_true, np.round(all_y_scores))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
plt.title("Overall Confusion Matrix")
plt.show()
```
